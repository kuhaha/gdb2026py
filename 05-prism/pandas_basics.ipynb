```python
df = pd.read_csv('golf_play.csv', encoding="utf8") 
df.head()
```

```python
df.tail()
```

```python
# ラベル指定：loc による指定列名や条件式で抽出。
# 天気予報Outlookが"rainy"である行のPlayの列を選択 
df.loc[df['Outlook'] == 'rainy', 'Play']
```

```python
# インデックス指定：iloc による指定行・列番号（数値）で指定。
# 0〜4行目、0〜2列目を抽出 
df.iloc[0:5, 0:2]
```

```python
# 複数条件：論理演算フィルタAND(&)やOR（|）で指定
# Outlookがsunnyで かつ Humidityがnormalである行 
df[(df['Outlook']=='sunny') & (df['Humidity']=='normal')]
```
